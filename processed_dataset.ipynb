```python
import pandas as pd
orders=pd.read_csv('/content/Day9_Orders.csv')
customers=pd.read_csv('/content/Day9_Customers.csv')
products=pd.read_csv('/content/Day9_Products.csv')
```

**Using concat() to demonstrate combining DataFrames**

```python
d=pd.concat([orders,customers,products],axis=1)
d
```

**Merging the files together**

```python
df=pd.merge(orders,customers,on="Customer_ID")
df
```

```python
df=df.merge(products,on="Product_ID")
df
```

**Using apply() to create Order_Value**

```python
df["Order_Value"] = df.apply(lambda row: row["Quantity"] * row["Unit_Price"],axis=1)
df.head()
```

**Using DateTime operations to convert the order date and extract information such as the month, day, or day of the week**

```python
df["Order_Date"].dtype
```

```python
df["Order_Date"]=pd.to_datetime(df["Order_Date"])
df["Order_Date"].dtype
```

```python
df["Order_Year"]=df["Order_Date"].dt.year
df["Order_Month"]=df["Order_Date"].dt.month
df["Order_Month_Name"]=df["Order_Date"].dt.month_name()
df["Order_Day"] =df["Order_Date"].dt.day
df["Order_Day_Name"] =df["Order_Date"].dt.day_name()
df.head()
```

**Export to CSV**

```python
df.to_csv("Processed_Ecommerce_Dataset.csv",index=False)
```
